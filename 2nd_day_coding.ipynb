```python
def calculator(a,b,operator):
    if operator == "addition":
        return a+b
    elif operator=="subtraction":
        return a-b
    elif operator=="multiplication":
        return a*b
    elif operator=="division":
        return a/b
    else:
        return None

[calculator(i,i+5,"multiplication") for i in range(10)]
```
